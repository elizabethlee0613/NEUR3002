```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

# --- Neuron Parameters ---
C = 200.0
V_th = -50.0
V_reset = -65.0
V_rest = -65.0
I_ext = 300.0
dt = 0.1

T = 100
time = np.arange(0, T, dt)

# --- Initialize voltage and spike times ---
V = np.full_like(time, V_rest)
spike_times = []

# --- Simulation ---
for i in range(1, len(time)):

    dV = (I_ext / C) * dt
    V[i] = V[i-1] + dV

    if V[i] >= V_th:
        V[i-1] = 40.0
        V[i] = V_reset
        spike_times.append(time[i])


# --- Voltage Plot ---
plt.figure(figsize=(10, 4))

plt.plot(time, V, label="Membrane Voltage")
plt.axhline(
    V_th,
    linestyle="--",
    label="Threshold Voltage"
)

plt.xlabel("Time (ms)")
plt.ylabel("Membrane Voltage (mV)")
plt.title("Integrate-and-Fire Model of Voltage Over Time")

plt.legend()
plt.grid(True)
plt.show()


# --- Raster Plot ---
plt.figure(figsize=(10, 2))

plt.eventplot(
    spike_times,
    lineoffsets=1,
    linelengths=0.8
)

plt.xlabel("Time (ms)")
plt.ylabel("Neuron")
plt.title("Raster Plot of Integrate-and-Fire Model Spike Times")

plt.yticks([1], ["Neuron 1"])
plt.xlim(0, T)
plt.ylim(0.5, 1.5)

plt.show()
```
